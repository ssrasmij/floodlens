```python
from pathlib import Path
import geopandas as gpd

# Find the FloodLens project directory
project_root = Path.cwd()

if not (project_root / "data").exists():
    project_root = project_root.parent

# Locate the floodplain dataset
flood_path = project_root / "data" / "raw" / "fema_floodplain_2023.geojson"

print("Dataset:", flood_path)
print("File exists:", flood_path.exists())

# Load GeoJSON into a GeoDataFrame
flood = gpd.read_file(flood_path)

print("Total flood features:", len(flood))
print("CRS:", flood.crs)
print("Geometry types:", flood.geom_type.unique())

display(flood.head())
```

```python
# Explore flood-zone classifications

print("=== FLOOD ZONES ===")
print(flood["fld_zone"].value_counts(dropna=False))

print("\n=== ZONE SUBTYPES ===")
print(flood["zone_subty"].value_counts(dropna=False))

print("\n=== SPECIAL FLOOD HAZARD AREAS ===")
print(flood["sfha_tf"].value_counts(dropna=False))

print("\n=== FLOOD ZONE COMBINATIONS ===")
display(
    flood.groupby(
        ["fld_zone", "zone_subty", "sfha_tf"],
        dropna=False
    ).size().reset_index(name="feature_count")
)
```

```python
# Load Philadelphia street centerlines

streets_path = (
    project_root / "data" / "raw" /
    "philadelphia_streets.geojson"
)

print("File exists:", streets_path.exists())

streets = gpd.read_file(streets_path)

print("\n=== STREET DATASET ===")
print("Total street features:", len(streets))
print("CRS:", streets.crs)
print("Geometry types:")
print(streets.geom_type.value_counts())

print("\n=== AVAILABLE COLUMNS ===")
print(streets.columns.tolist())

print("\n=== FIRST FIVE RECORDS ===")
display(streets.head())
```

```python
# Stage 2.4.5 - Check street data quality

print("=== STREET IDENTIFIERS ===")
print("Total records:", len(streets))
print("Unique segment IDs:", streets["seg_id"].nunique())
print("Missing segment IDs:", streets["seg_id"].isna().sum())
print("Duplicate segment IDs:", streets["seg_id"].duplicated().sum())

print("\n=== STREET NAMES ===")
print("Missing street labels:", streets["streetlabe"].isna().sum())
print(
    "Blank street labels:",
    streets["streetlabe"].fillna("").str.strip().eq("").sum()
)

print("\n=== GEOMETRY QUALITY ===")
print("Missing geometries:", streets.geometry.isna().sum())
print("Empty geometries:", streets.geometry.is_empty.sum())
print("Invalid geometries:", (~streets.geometry.is_valid).sum())

print("\n=== SAMPLE STREET RECORDS ===")
display(streets[["seg_id", "streetlabe", "class", "geometry"]].head(10))
```
